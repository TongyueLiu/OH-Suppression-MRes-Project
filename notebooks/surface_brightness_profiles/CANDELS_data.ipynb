```python
# Your input coordinate in decimal degrees
ra_deg = 53.15275
dec_deg = -27.7505

# Try numeric-degree matching first
mask = (
    np.isclose(df_t2["RAJ2000"], ra_deg, atol=1e-4) &
    np.isclose(df_t2["DEJ2000"], dec_deg, atol=1e-4)
)

# Fallback: if the columns are stored as strings, try exact string matching
if mask.sum() == 0:
    mask = (
        df_t2["RAJ2000"].astype(str).str.strip() == "03:32:36.66"
    ) & (
        df_t2["DeJ2000"].astype(str).str.strip() == "-27:45:01.8"
    )

matches = df_t2.loc[mask]

print(f"Matched rows: {len(matches)}")
print(matches)
```
