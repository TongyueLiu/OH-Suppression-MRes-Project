```python
import numpy as np
from matplotlib import pyplot as plt
from scipy import * 
# from sympy import *
```

Fresnel's equations: 

$$
r_s = -\frac{sin(\Theta_i - \Theta_t)}{sin(\Theta_i + \Theta_t)}
$$

$$
r_p = \frac{tan(\Theta_i - \Theta_t)}{tan(\Theta_i + \Theta_t)}
$$

where $r_s$ and $r_p$ are the amplitude reflection coefficient. 

The percentage of power reflected back is: 


$R_s = |r_s|^2$, and $R_p = |r_p|^2$


Total Reflectance is given by: 
$$
R = \frac{R_s + R_p}{2}
$$

```python
# refractive index and extinction coefficient of silver at 1.50 microns
n_Ag = 0.13986 + 10.963j
# k_Ag = 10.963

# refractive index and extinction coefficient of aluminium at 1.50 microns
n_Al = 1.5237 + 15.115j
# k_Al = 15.115
```

```python
# Incidence angle of each mirror in degrees
theta_M3_i = 45.0 
theta_M4_i = 45.0 
theta_M5_i = 4 
theta_M6_i = 25 
theta_M7_i = 7 
theta_M8_i = 13 
theta_M9_i = 45
```

Approximate the angle of transmittance using the snell's law: 
$$
n_1 sin\theta_i = n_2 sin \theta_t
$$

Assume incident medium is air, so n1 = 1. 

So $sin\theta_i = n sin \theta_t$

$$
\theta_t = sin^{-1}\left(\frac{sin\theta_i}{\~n_2}\right)
$$

```python
# Assume incident medium is air, so n1 = 1. 

# Aluminium mirror M3 inclined at 45 degrees: 

def find_reflectivity(theta_i_deg, n_2, n_1 = 1.0): 
    """
    Calculate the total reflectivity R = 1/2(R_s + R_p)

    Parameters 
    ------
    theta_i_deg: the angle of incidence at the air-coating medium, in degrees

    n_2: the refractive index of the coating, includes complex component

    n_1: refractive index of the incident medium. Default is air, n_1 = 1
    """

    # convert angle from degrees to radians
    theta_i = np.deg2rad(theta_i_deg)

    theta_t = np.emath.arcsin(n_1 * np.sin(theta_i) / n_2)
    r_s = - np.sin(theta_i - theta_t) / np.sin(theta_i + theta_t)
    r_p = np.tan(theta_i - theta_t) / np.tan(theta_i + theta_t)

    R_s = (np.abs(r_s))**2
    R_p = (np.abs(r_p))**2 

    R_tot = 0.5 * (R_s + R_p)

    return R_tot, R_s, R_p
```

```python
R_M3 = find_reflectivity(theta_M3_i, n_Al)[0]
R_M4 = find_reflectivity(theta_M4_i, n_Ag)[0]
R_M5 = find_reflectivity(theta_M5_i, n_Ag)[0]
R_M6 = find_reflectivity(theta_M6_i, n_Ag)[0]
R_M7 = find_reflectivity(theta_M7_i, n_Ag)[0]
R_M8 = find_reflectivity(theta_M8_i, n_Ag)[0]
```

```python
import numpy as np

def find_reflectivity_cosine(theta_i_deg, n_2, n_1=1.0):
    """
    Unpolarized reflectivity of a single interface.

    Parameters
    ----------
    theta_i_deg : float or array-like
        Incidence angle in degrees, measured from the surface normal.
    n_2 : complex
        Complex refractive index of the second medium.
    n_1 : complex, optional
        Refractive index of the incident medium.
    """

    theta_i = np.deg2rad(theta_i_deg)

    n_1 = complex(n_1)
    n_2 = complex(n_2)

    sin_theta_t = n_1 * np.sin(theta_i) / n_2

    cos_theta_t = np.lib.scimath.sqrt(
        1.0 - sin_theta_t**2
    )

    # Select the passive/decaying transmitted-wave branch.
    kz_2 = n_2 * cos_theta_t

    if np.imag(kz_2) < 0:
        cos_theta_t = -cos_theta_t

    cos_theta_i = np.cos(theta_i)

    r_s = (
        n_1 * cos_theta_i - n_2 * cos_theta_t
    ) / (
        n_1 * cos_theta_i + n_2 * cos_theta_t
    )

    r_p = (
        n_2 * cos_theta_i - n_1 * cos_theta_t
    ) / (
        n_2 * cos_theta_i + n_1 * cos_theta_t
    )

    R_s = np.abs(r_s)**2
    R_p = np.abs(r_p)**2
    R_tot = 0.5 * (R_s + R_p)

    return R_tot, R_s, R_p
```
