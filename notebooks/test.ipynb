```python
from joblib import Parallel, delayed

def square(x):
    return x**2

result = Parallel(n_jobs=2)(
    delayed(square)(x)
    for x in range(10)
)

print(result)
```

```python
from joblib import Parallel, delayed
from scipy.special import jv

def test_bessel(x):
    return jv(0, x)

result = Parallel(n_jobs=2)(
    delayed(test_bessel)(x)
    for x in range(10)
)

print(result)
```
